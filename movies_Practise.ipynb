```python
# Write a small function that classifies runtime into 'Short' (<90 mins), 'Standard' (90-130 mins), and 'Long' (>130 mins). Apply this function to create a new column duration_type.
def runtimeclass(val):
  if(val < 90):
    return 'Short'
  elif((val >= 90) & (val <= 130)):
    return 'Standard'
  else:
    return 'Long'

df['duration_type'] = df['runtime'].apply(runtimeclass)
df
```

```python
# 35. Cumulative Calculations: Create a new column cumulative_gross_by_director. This should show the running total (cumulative sum) of gross for each director, ordered by year.

# First sort by year so the 'running total' makes chronological sense
df = df.sort_values('year')
df['cumulative_gross_by_director'] = df.groupby('director')['gross'].cumsum()
df['cumulative_gross_by_director']
```

```python
# 36. The apply logic (Multi-column): Create a new column is_blockbuster. A movie is a blockbuster (True/False) if its gross is more than 3 times its budget AND its score is higher than 7.0. Write a function and use .apply() with axis=1.

def check_blockbuster(row):
    if (row['gross'] > (3 * row['budget'])) and (row['score'] > 7.0):
        return True
    else:
        return False

# axis=1 tells pandas to feed the function one row at a time
df['is_blockbuster'] = df.apply(check_blockbuster, axis=1)
```

```python
# 37. Date Extractions: From your release_date column (ensure it's datetime), create two new columns: release_year (as an integer) and is_weekend (a boolean True/False if the movie was released on a Saturday or Sunday).

# Ensure it is a datetime object first
df['release_date'] = pd.to_datetime(df['release_date'])

df['release_year'] = df['release_date'].dt.year
# .dt.dayofweek returns 5 for Saturday and 6 for Sunday
df['is_weekend'] = df['release_date'].dt.dayofweek.isin([5, 6])
```
