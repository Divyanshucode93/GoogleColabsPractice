```python
# The released column is currently a string. Create a new column release_date that is a proper datetime object. (Hint: You'll need to strip the country name first. You can use .str.replace or regex to remove everything from the first ( to the end).
df['release_date'] = df['released'].str.split('(').str.get(0).str.strip()
df
```

```python
# We noticed 1392 movies have a budget of 0. This is bad for analysis. Replace those 0 values with the median budget of their respective genre. (Hint: Use groupby + transform).
df.groupby('genre')['budget'].fillna(value=df.groupby('genre')['budget'].median())
```

```python
# Create a new categorical column score_category where:
# Score < 5.0 --> 'Low'
# 5.0 <= Score < 7.5 --> 'Medium'
# Score >= 7.5 --> 'High'
def scorecat(val):
  if val < 5.0:
    return 'Low'
  elif (val >= 5.0) & (val < 7.5):
    return 'Medium'
  else:
    return 'High'

df['score_category'] = df['score'].apply(scorecat)
df
```
