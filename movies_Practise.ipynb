```python
# The released column is currently a string. Create a new column release_date that is a proper datetime object. (Hint: You'll need to strip the country name first. You can use .str.replace or regex to remove everything from the first ( to the end).
df['release_date'] = df['released'].str.split('(').str.get(0).str.strip()
df
```

```python
# We noticed 1392 movies have a budget of 0. This is bad for analysis. Replace those 0 values with the median budget of their respective genre. (Hint: Use groupby + transform).
df.groupby('genre')['budget'].fillna(value=df.groupby('genre')['budget'].median())
```

```python
# Create a new categorical column score_category where:
# Score < 5.0 --> 'Low'
# 5.0 <= Score < 7.5 --> 'Medium'
# Score >= 7.5 --> 'High'
def scorecat(val):
  if val < 5.0:
    return 'Low'
  elif (val >= 5.0) & (val < 7.5):
    return 'Medium'
  else:
    return 'High'

df['score_category'] = df['score'].apply(scorecat)
df
```

```python
# Find the top 5 director and star combinations that have the highest average score, but only include pairs that have made at least 3 movies together.
duos = df.groupby(['director', 'star'])['score'].agg(['mean', 'count'])
top_duos = duos[duos['count'] >= 3].sort_values('mean', ascending=False).head(5)
top_duos
```

```python
# Using the release_date column you created in Q1, find which month of the year (January-December) has the highest average gross earnings.
df['release_date'] = pd.to_datetime(df['released'].str.split('(').str.get(0).str.strip(), errors='coerce')
df
```

```python
# Calculate what percentage of the total gross of the entire dataset is contributed by the top 5 companies.
total_gross = df['gross'].sum()
top_5_companies_gross = df.groupby('company')['gross'].sum().sort_values(ascending=False).head(5).sum()
perc = (top_5_companies_gross / total_gross) * 100
perc
```

```python
# Write a small function that classifies runtime into 'Short' (<90 mins), 'Standard' (90-130 mins), and 'Long' (>130 mins). Apply this function to create a new column duration_type.
def runtimeclass(val):
  if(val < 90):
    return 'Short'
  elif((val >= 90) & (val <= 130)):
    return 'Standard'
  else:
    return 'Long'

df['duration_type'] = df['runtime'].apply(runtimeclass)
df
```

```python
# Basic Aggregation: Group the movies by rating. Find the count of movies and the average score for each rating.
avg_count = df.groupby('rating')[['name','score']].agg({'name':'count','score':'mean'})
avg_count
```

```python
# Named Aggregation: Perform the same operation as above, but name the resulting columns total_movies and average_movie_score.

# This creates a NEW summary DataFrame
rating_summary = df.groupby('rating').agg(
    total_movies=('name', 'count'),
    average_movie_score=('score', 'mean')
)
print(rating_summary)
df
```

```python
# Handling Zeroes: Replace all 0 values in the budget column with NaN. Then, count how many NaN values are now in that column.

# You must assign it back to the column to "save" the NaNs
df['budget'] = df['budget'].replace(0, np.nan)

# Now you can count them
print(df['budget'].isna().sum())
```

```python
# Transform (Intermediate): Create a new column genre_avg_gross. This column should contain the mean gross of the genre that the movie belongs to. (Hint: Use groupby('genre')['gross'].transform('mean')).

# .transform() calculates the mean but keeps the original shape (4,000 rows)
df['genre_avg_gross'] = df.groupby('genre')['gross'].transform('mean')
df
```

```python
# Filtering Groups: Use groupby and filter to find all rows belonging to genres that have fewer than 10 movies in the entire dataset.
df.groupby('genre').filter(lambda x: len(x) < 10)
```

```python
# Multi-Column Grouping: Calculate the total gross and total budget for every combination of year and genre.
df.groupby(['year','genre'])[['gross','budget']].sum().reset_index()
```

```python
# Data Cleaning & Grouping: Some company names might have leading or trailing spaces. Strip the spaces from the company column first, then count how many movies each company has produced.
df['company'] = df['company'].str.strip()
df.groupby('company')['name'].count()
```

```python
# Percentage Calculation: Group by genre and calculate the total votes for each. Then, calculate what percentage of the total votes (of all movies) each genre represents.
# 1. Total votes across the whole dataset
grand_total_votes = df['votes'].sum()

# 2. Total votes per genre divided by grand total
percentage_votes = (df.groupby('genre')['votes'].sum() / grand_total_votes) * 100

percentage_votes.sort_values(ascending=False)
```
