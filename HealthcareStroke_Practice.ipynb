# **Set 3 (Questions 21-30)**
**Advanced Filtering & Logic**

```python
# Sort the DataFrame first by gender (ascending) and then by age (descending) in a single command.
df.sort_values(by=['gender','age'],ascending=[True,False])
```

```python
# Use the .rename() method to change bmi to Body_Mass_Index and stroke to Stroke_Result simultaneously.
df.rename(columns={'bmi':'Body_Mass_Index','stroke':'Stroke_Result'},inplace=True)
```

```python
# Calculate the percentage of the total population that is "Male".
perc = ((df['gender']=='Male').sum()) / (df.shape[0]) * 100
perc
```

```python
# Filter the data to find patients who are "Private" workers AND have an avg_glucose_level greater than 200.
df[(df['work_type']=='Private') & (df['avg_glucose_level']>200)]
```

```python
# Find the average age of people who have had a stroke vs. those who haven't (Use groupby).
df.groupby('Stroke_Result')['age'].mean()
```

```python
# Create a column named is_obese which is True if bmi is greater than 30, and False otherwise.
df['is_obese'] = df['Body_Mass_Index'] > 30
df.head()
```

```python
# Count how many unique values exist in the work_type column.
df['work_type'].nunique()
```

```python
# Retrieve the 5th to 15th rows of the DataFrame using .iloc.
df.iloc[4:15]
```

```python
# Filter for patients who are either "formerly smoked" or "smokes". (Hint: try using .isin()).
df[df['smoking_status'].isin(['formerly smoked','smokes'])]
```

```python
# Fill all the missing values in the bmi column with the median value of that column.
df['Body_Mass_Index'].fillna(value=df['Body_Mass_Index'].median(),inplace=True)
df
```
